```python
career_count = df["career"].value_counts()

career_count.plot(kind="bar")

plt.title("Career Distribution")
plt.xlabel("Career")
plt.ylabel("Count")
plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(df["career"], df["starting_salary_LPA"])

plt.xticks(rotation=90)

plt.xlabel("Career")

plt.ylabel("Starting Salary (LPA)")

plt.title("Starting Salary of Different Careers")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(df["career"], df["salary_5yr_LPA"])

plt.xticks(rotation=90)

plt.xlabel("Career")

plt.ylabel("Salary after 5 Years (LPA)")

plt.title("5-Year Salary Comparison")

plt.show()
```
